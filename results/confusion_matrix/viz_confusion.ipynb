```python
import json, numpy as np, plotly.graph_objects as go

with open("test_epoch_000_digits.json") as f:
    data = json.load(f)

cm = np.array(data["matrix"], dtype=float)
labels = data["class_names"]

# row-normalize to percentages
cm_pct = cm / cm.sum(axis=1, keepdims=True) * 100
text = [[f"{v:.1f}" for v in row] for row in cm_pct]

fig = go.Figure(go.Heatmap(
    z=cm_pct, x=labels, y=labels,
    colorscale="Blues",
    text=text,
    texttemplate="%{text}%",
    textfont_size=11,
))
fig.update_xaxes(dtick=1)
fig.update_yaxes(dtick=1, autorange="reversed")
fig.update_layout(
    xaxis_title="Predicted",
    yaxis_title="True",
    # title=f"Confusion Matrix: Emnist Digits (Conv)",
    width=500, height=550,
)
fig.show(config={"toImageButtonOptions": {"format": "svg"}})
```

```python
import json, numpy as np, plotly.graph_objects as go
from plotly.subplots import make_subplots

blues = [
    [0.0, "#f7fbff"], [0.125, "#deebf7"], [0.25, "#c6dbef"],
    [0.375, "#9ecae1"], [0.5, "#6baed6"], [0.625, "#4292c6"],
    [0.75, "#2171b5"], [0.875, "#08519c"], [1.0, "#08306b"],
]

paths = [
    "ssl_emnist_digits_conv_cm.json",
    "ssl_emnist_digits_clip-5_cm.json",
    "ssl_emnist_digits_dino-3_cm.json",
]
titles = ["EMNIST Digits Conv", "EMNIST Digits CLIP", "EMNIST Digits DINO"]

# load and normalize
cms, labels = [], None
for p in paths:
    with open(p) as f:
        d = json.load(f)
    cm = np.array(d["matrix"], dtype=float)
    cms.append(cm / cm.sum(axis=1, keepdims=True) * 100)
    labels = d["class_names"]

vmin = min(c.min() for c in cms)
vmax = max(c.max() for c in cms)

fig = make_subplots(rows=1, cols=3, subplot_titles=titles, horizontal_spacing=0.08)

for i, cm_pct in enumerate(cms):
    text = [[f"{v:.1f}" for v in row] for row in cm_pct]
    fig.add_trace(go.Heatmap(
        z=cm_pct, x=labels, y=labels,
        colorscale=blues, zmin=vmin, zmax=vmax,
        text=text, texttemplate="%{text}%", textfont_size=9,
        showscale=(i == len(cms) - 1),  # colorbar on last only
    ), row=1, col=i + 1)

fig.update_xaxes(dtick=1)
fig.update_yaxes(dtick=1, autorange="reversed")
fig.update_layout(width=1400, height=500)
fig.show(config={"toImageButtonOptions": {"format": "svg"}})
```
